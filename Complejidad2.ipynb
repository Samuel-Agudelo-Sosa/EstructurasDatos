```python
def algoritmo(n):
    cnt = 0
    i = 1
    cnt += 1
    while i <= n:
        cnt += 1
        k = i 
        cnt += 1
        while k <= n:
            cnt += 1
            k += 2
            cnt += 1
        cnt += 1
        k = 1
        cnt += 1
        while k <= i:
            cnt += 1
            k += 1
            cnt += 1
        cnt += 1
        i *= 2
        cnt += 1
    cnt += 1
    return cnt
```

```python
import math
def complejidadTeorica(n):
    l1 = 1
    l2 = math.floor(math.log2(n)) + 2
    l3 = math.floor(math.log2(n)) + 1
    l4 = math.ceil(n/2) + 2 +  math.floor(math.log2(n))  * (math.floor(n / 2) + 2)  - 2 ** math.floor(math.log2(n))
    l5 = l4 - math.floor(math.log2(n)) - 1
    l6 = l3
    l7 = 2 ** (math.floor(math.log2(n)) + 1) + math.floor(math.log2(n))
    l8 = 2 ** (math.floor(math.log2(n)) + 1) - 1
    l9 = l3
    return l1 + l2 + l3  + l4 + l5 + l6 + l7 + l8 + l9  
```

```python
for i in range(1, 21):
    print("n: ", i, " Complejidad Temporal Real: ", algoritmo(i), " Complejidad Temporal Teorica: ", complejidadTeorica(i))
```
