```python
def T(n):
    if n == 1:
        return 1
    else:
        return n + 3 * T(n / 4)
```

```python
import math
def recurrenciaTeorica(n):
    return 4 * n - 3 * n ** (math.log(3, 4))
```

```python
lst = [4 ** x for x in range(1, 11)]

for n in lst:
    print(f"n: {n}, T(n): {T(n)}, Recurrencia Teórica: {recurrenciaTeorica(n)}")
```
