# Sumatorios

- Observemos que podríamos usar un bucle for o fórmula cerrada para resolver el siguiente sumatorio:

$$
\sum_{i=1}^{n} i
$$

```python
n = 100
suma = 0
for i in range(1, n+1):
    suma = suma + i

print("Se verifica que es igual a la fórmula cerrada " + str(suma) + " = " + str(n * (n + 1) // 2))
```

```python
k = 5
for i in range(5):
    for j in range(4):
        k = k + i - j
print(k)
```

```python
def insertionSort(l):
    for j in range(1,len(l)):
        key = l[j]
        posicion = j-1
        while posicion > -1 and l[posicion] > key:
            l[posicion + 1] = l[posicion]
            posicion = posicion - 1
        l[posicion + 1] = key
    return l

print(insertionSort([5,2,4,6,1,3,7,8,9,10,3,9,16]))
```
