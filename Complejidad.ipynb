```python
def algorithm1(n):
    c = 0
    
    i = 1
    c += 1 # L1
    
    while i <= n:
        c += 1 # L2 (Comprobación Verdadera)
        
        k = i
        c += 1 # L3
        
        while k <= n:
            c += 1 # L4 (Comprobación Verdadera)
            k = k + 2
            c += 1 # L5
        c += 1 # L4 (Comprobación Falsa - salida del bucle)
        
        k = 1
        c += 1 # L6
        
        while k <= i:
            c += 1 # L7 (Comprobación Verdadera)
            k = k + 1
            c += 1 # L8
        c += 1 # L7 (Comprobación Falsa - salida del bucle)
        
        i = i + 2
        c += 1 # L9
        
    c += 1 # L2 (Comprobación Falsa - salida del bucle principal)
    
    return c
```

```python
import math 
def ComplejidadTeorica(n):
    m = math.ceil(n / 2) # Simplifico esto en una variable para que sea más legible
    
    l1 = 1
    l2 = m + 1
    l3 = m 
    l4 = (m * (m + 3)) / 2 
    l5 = (m * (m + 1)) / 2 
    l6 = m
    l7 = m * (m + 1)
    l8 = m * m
    l9 = m 
    
    return l1 + l2 + l3 + l4 + l5 + l6 + l7 + l8 + l9
```

```python
for i in range(1, 11):
    print("n: ", i, " Complejidad Temporal Real: ", algorithm1(i), " Complejidad Temporal Teorica: ", ComplejidadTeorica(i))
```
