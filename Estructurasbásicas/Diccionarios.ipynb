# Diccionarios 

Los diccionarios son una estructura de dato cuyos elementos son pares clave/valor
- **Ordenados** Los diccionarios están indexados por la clave y está debe ser un tipo inmutable por ejemplo strings o números.
- **Unicidad Claves:** No puede tener dos claves iguales. Si intentas asignar un nuevo valor a una clave ya existente, el valor antiguo se sobreescribe.
- **Mutable:** Se puede añadir eliminar o modificar los valores del diccionario.
- **Claves hashables:** Sus claves deben ser inmutables como strings, enteros, booleanos o tuplas (si sus elementos son inmutables también), listas no serviría de hecho.

```python
#creación de un diccionario usando corchetes
diccionario1 = {'clave1': 'valor1', 'clave2': 'valor2', 'clave3': 'valor3'}
print(diccionario1)

#creación de un diccionario usando el constructor dict()
diccionario2 = dict(clave1='valor1', clave2='valor2', clave3='valor3')
print(diccionario2)
```

- Se puede indexar mediante claves y se pueden añadir o modificar un elemento, especificando la clave, si ya existe se modifica el elemento con dicha clave y un nuevo valor, si no existe entonces se crea un nuevo elemento con la clave y valor especificados. Para elminar se usa del.

```python
persona = {
    'nombre': 'Juan',
    'edad': 30,
    'ciudad': 'Madrid'
}

#Indexando elementos
print(persona['nombre']) #accediendo al valor asociado a la clave 'nombre', si la clave no existe, se lanzará un KeyError
print(persona.get('nombre')) # otra forma de acceder a los valores, aunque con get() si la clave no existe, devuelve None en lugar de lanzar un error
#Añadiendo un nuevo elemento
persona['profesión'] = 'Ingeniero'
print(persona)  
#Modificando un elemento existente
persona['edad'] = 31
print(persona) 
del persona['ciudad']#Borrando un elemento con del
print(persona.pop('profesión'))#Borrando un elemento con pop(), devuelve el valor asociado a la clave eliminada
print(persona)
persona.clear()#Borrando todos los elementos del diccionario
print(persona)
```

- tenemos != y = que verifican si dos diccionarios son iguales, y in y not in que actuan sobre las claves

```python
diccionarioA = {'clave1': 'valor1', 'clave2': 'valor2'}
diccionarioB = {'clave1': 'valor1', 'clave2': 'valor2'}
#Comparando diccionarios
print(diccionarioA == diccionarioB)  # True, porque tienen el mismo contenido
print(diccionarioA != diccionarioB)  # False, porque tienen el mismo contenido
#Uso de in y not in
print('clave1' in diccionarioA)  # True, porque 'clave1' es una clave en diccionarioA
print('clave3' in diccionarioA)  # False, porque 'clave3' no es una clave en diccionarioA
print('clave3' not in diccionarioA)  # True, porque 'clave3' no es una clave en diccionarioA
```

- Tenemos los métodos Keys que devuelve una vista de todas las claves, values que devuelve una vista de todos lso valores, items que devuelve parejas en forma de tuplas, update que actualiza el diccionario con otros pares clave-valor.
