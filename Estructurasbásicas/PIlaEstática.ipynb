# TAD Pila (Estructura de Datos LIFO)

## Concepto Principal
La **pila** es una estructura de datos que opera bajo el principio **LIFO** (*Last In, First Out*). Esto significa estrictamente que **los últimos elementos en entrar a la estructura serán siempre los primeros en salir**. Por su naturaleza restrictiva, no es posible acceder a ningún elemento de la estructura que no sea el último que fue añadido.

## Tipos de Implementación
Dependiendo de si se establece o no un límite de elementos, las pilas pueden ser:
* **Pilas Estáticas:** Se implementan utilizando contenedores de tamaño fijo (como *arrays*). Requieren establecer un tamaño máximo desde el momento de su creación.
* **Pilas Dinámicas:** Se implementan utilizando el concepto de listas enlazadas, donde existen nodos conectados por punteros. Crecen y decrecen según sea necesario.

## Operaciones Básicas
El comportamiento abstracto de una pila se define mediante las siguientes operaciones:

* **Inicializar():** Genera una pila nueva y vacía. Si la pila ya existía y contenía elementos, estos se eliminan.
* **Apilar(Elemento):** Añade un nuevo elemento a la cima de la pila. En implementaciones estáticas, esta operación puede fallar (devolviendo `False` o un error) si se ha alcanzado la capacidad máxima y `True` si la operación se ha realizado con éxito.
* **Desapilar():** Obtiene y **elimina** el último elemento que entró a la pila. Es la operación de extracción por defecto.
* **Cima() o Tope():** Obtiene una referencia al dato que está en la cima (el último ingresado), pero **sin borrarlo** de la estructura. 
* **Vacia():** Función de comprobación que devuelve un valor booleano (`True` o `False`) para indicar si la pila tiene o no elementos. Es fundamental ejecutarla antes de intentar desapilar para evitar errores.
* **Limpiar():** Elimina todos los elementos de la pila de una sola vez, desapilando todos sus elementos y dejándola completamente vacía.

## Lógica Interna de Control (Enfoque Estático)
En la teoría de su implementación estática, el control de los elementos se gestiona mediante un indicador llamado **tope**:

1.  **Estado Inicial:** El `tope` inicia en un valor que indica que no hay elementos (generalmente `-1`).
2.  **Inserción (Apilar):** Se desplaza el `tope` a la siguiente posición disponible y luego se inserta el elemento en dicha posición.
3.  **Extracción (Desapilar) / Consulta (Cima):** Siempre se valida que la pila no esté vacía (`tope < 0`). Si está vacía, se debe lanzar una **excepción** para informar del error. Si hay datos, la función devuelve el elemento apuntado por el `tope`. Si la operación es *desapilar*, adicionalmente se decrementa el `tope` para marcar el nuevo límite de la pila. 

## Casos de Uso y Aplicaciones de las Pilas

Las estructuras de datos tipo pila son fundamentales cuando **el orden de las acciones es importante**. Su naturaleza LIFO (Último en entrar, Primero en salir) asegura que un sistema no pase a una acción anterior sin haber completado o descartado las más recientes. 

A continuación, se detallan sus aplicaciones más comunes:

### Aplicaciones Principales (Basadas en el orden de ejecución)

* **Inversión de secuencias (Reversing):** De forma predeterminada, una pila invertirá cualquier secuencia de datos que se ingrese. 
    * *Ejemplo:* Para invertir la cadena "Hola", apilamos cada carácter ('H', 'o', 'l', 'a') secuencialmente. Al desapilarlos, saldrán en orden inverso ('a', 'l', 'o', 'H').
* **Funcionalidad Deshacer/Rehacer (Undo/Redo):** Utilizado en editores de texto, software de diseño, etc. El estado del programa (o la acción realizada) se apila cada vez que se hace un cambio. Para "deshacer", simplemente se desapila el último cambio. 
    * *Nota adicional:* Para implementar un sistema completo de "Rehacer", habitualmente se utilizan **dos pilas**: una guarda el historial de acciones y la otra guarda las acciones deshechas.
* **Algoritmos de Retroceso (Backtracking):** Esencial para algoritmos de resolución de problemas que implican explorar múltiples caminos (como salir de un laberinto, el problema de las N-Reinas o el recorrido de grafos mediante Búsqueda en Profundidad - DFS). 
    * *Mecánica:* Cada vez que se elige un nodo o ruta, se apila. Si la ruta resulta ser un callejón sin salida, se extrae esa última decisión (`pop()`) y se intenta con la siguiente ruta disponible.
* **Pila de llamadas de ejecución (Call Stack):** Es el mecanismo interno que usan los lenguajes de programación para gestionar la ejecución de funciones. 
    * *Funcionamiento:* Cuando se llama a una función, su contexto (variables locales, dirección de retorno) se agrega a la pila de llamadas. Cuando la función termina su ejecución, su contexto se desapila y el programa sabe exactamente a qué punto de la función anterior debe regresar.

* **Verificación de paréntesis y sintaxis:** Los compiladores y editores de código usan pilas para verificar si expresiones matemáticas o bloques de código tienen los paréntesis, llaves y corchetes correctamente balanceados y cerrados.
* **Historial del navegador web:** Los botones de "Atrás" y "Adelante" del navegador funcionan con pilas. Al visitar una página nueva, la URL actual se apila en la pila de "Atrás". Al presionar el botón "Atrás", se desapila esa URL para visitarla y la página actual pasa a la pila de "Adelante".
* **Evaluación y conversión de expresiones:** Las calculadoras y compiladores utilizan pilas para convertir expresiones matemáticas de notación infija (ej. `3 + 4 * 2`) a notación postfija o polaca inversa (ej. `3 4 2 * +`), y posteriormente usar otra pila para evaluar el resultado exacto respetando la jerarquía de operadores.

# Implementación de las pilas estáticas en Python.

Se añade la clase pila estática:

```python
class PilaEstatica:
    def __init__(self, tamaño):
        self.tamaño = tamaño
        self.inicializar()

    def inicializar(self):
        self.pila = [0] * self.tamaño
        self.tope = -1

    def limpiar(self):
        self.tope = -1

    def apilar(self, elemento):
        if self.tope < self.tamaño - 1:
            self.tope += 1
            self.pila[self.tope] = elemento 
            return True
        return False
    
    def cima(self):
        if self.tope < 0:
            raise Exception("La pila está vacía")
        return self.pila[self.tope]
    
    def desapilar(self):
        if self.tope < 0:
            raise Exception("La pila está vacía")
        valor = self.pila[self.tope]
        self.tope -= 1
        return valor  
    
    def vacia(self):
        return self.tope < 0
    
    def __str__(self):
        if self.vacia():
            return "Pila vacía"
        return "Pila: " + ", ".join(str(self.pila[i]) for i in range(self.tope + 1))
    
    def __repr__(self):
        return self.__str__()
    
  
```

Ejemplos de uso:

```python
mi_pila = PilaEstatica(4)

print(mi_pila.vacia())  # True

mi_pila.apilar(1)
mi_pila.apilar(2)
mi_pila.apilar(3)
mi_pila.apilar(4)
print(mi_pila)
print(mi_pila.apilar(5))  # False, la pila está llena
print(mi_pila.cima())  # 4
print(mi_pila.desapilar())  # 4
print(mi_pila.cima())  # 3
print(mi_pila.vacia())  # False
mi_pila.limpiar()
print(mi_pila.vacia())  # True
```

Ejemplo de aplicación invertir lista:

```python
def invertir_cadena(cadena):
    mi_pila = PilaEstatica(len(cadena))
    for caracter in cadena:
        mi_pila.apilar(caracter)
    cadena_invertida = ""
    
    while not mi_pila.vacia():
        cadena_invertida += mi_pila.desapilar()

    return cadena_invertida

print(invertir_cadena("HOLA"))

    
```

Vaciar pila recursivamente:

```python
def vaciar_pila(pila):
    if pila.vacia():
        print("La pila ya está vacía")
    else:
        pila.desapilar()
        vaciar_pila(pila)

pila_ejemplo = PilaEstatica(5)
pila_ejemplo.apilar(10)
pila_ejemplo.apilar(20)
print(pila_ejemplo.vacia())  # False
vaciar_pila(pila_ejemplo)
print(pila_ejemplo.vacia())  # True
```

- Transferir pila invertida:

```python
def transferirPila(S, T):
    while not S.vacia():
        T.apilar(S.desapilar())

pila_S = PilaEstatica(5)
pila_T = PilaEstatica(5)
pila_S.apilar(1)
pila_S.apilar(2)
transferirPila(pila_S, pila_T)
print(pila_S.vacia())  # True
print(pila_T)
```
