# Listas

- Estructura de datos:
- **mutable:** la lista puede modificarse
- **ordenadas** los elementos son accedidos por índice.
- **heretogenea:** los elementos pueden ser de cualquier tipo (incluso listas)
- **elementos repetidos permitidos**



```python
lista = [1, 2, 3, 4, 4, 5, [6, 7], True, 0.15, [], "Samuel"] # se observa que puede ser heterogenea, es decir, puede contener diferentes tipos de datos y tener elementos repetidos
print(lista)
```

```python
lista[0] = 10 # Modificar el primer elemento
print(lista)
```

- otra alternativa para crear listas es usar el constructor de la clase list que tiene como argumento un **iterable** (listas, rangos, strings, tuplas, etc)

```python
lista = list(lista) # Crear una nueva lista a partir de otra lista existente, en este caso, se crea una copia de la lista original
print(lista)

```

```python
lista = list() # Crear una lista vacía utilizando el constructor de lista
print(lista)
```

```python
lista = list(range(0,10,2)) # Crear una lista a partir de un rango de números, en este caso, del 0 al 9 con un paso de 2
print(lista)
```

```python
lista = list("hola mundo") # Crear una lista a partir de una cadena de texto
print(lista)
```

```python
lista = list((1, 2))
print(lista)
```

```python
lista = list(list()) # Crear una lista a partir de otra lista vacía, lo que resulta en otra lista vacía ya que la lista original no contiene ningún elemento
print(lista)
```

```python
lista = [list()] # Crear una lista que contiene otra lista vacía como su único elemento, lo que resulta en una lista con un solo elemento que es otra lista vacía
print(lista)
```

```python
lista = list([5]) 
print(lista)
```

- para conocer el tamaño de una lista se utiliza **len()**. el último elemento de la lista será len()-1. También se pueden usar índices negativos,
el valor -1 permite acceder al último elemento, -(len) al primer elemento.

```python
lista = list(range(0,10, 3))
print(len(lista)) # Obtener la longitud de la lista, es decir, el número de elementos que contiene
print(lista[len(lista)-1]) # Acceder al último elemento de la lista utilizando la función len() para obtener el índice del último elemento
```

```python
print(lista[-1]) # Acceder al último elemento de la lista utilizando un índice negativo, en este caso, -1 se refiere al último elemento de la lista
print(lista[-2]) # Acceder al penúltimo elemento de la lista utilizando un índice negativo, en este caso, -2 se refiere al penúltimo elemento de la lista
print(-len(lista)) # Obtener el índice negativo que corresponde al primer elemento de la lista, en este caso, -len(lista) se refiere al primer elemento de la lista
```

- las listas se segmentan haciendo uso del operador inicio:fin este permite indicar donde empieza y donde termina (sin incluirlo) también está inicio:fin:pasos

```python
print(lista[2:5]) #se obtiene sublista desde el índice 2 hasta el índice 4, es decir, los elementos en las posiciones 2, 3 y 4 de la lista original
print(lista[:2]) #se obtiene sublista desde el inicio de la lista hasta el índice 4, es decir, los elementos en las posiciones 0, 1, 2, 3 y 4 de la lista original
print(lista[2:]) #se obtiene sublista desde el índice 2 hasta el final de la lista, es decir, los elementos en las posiciones 2, 3, 4 y 5 de la lista original
print(lista[:]) #se obtiene una copia de la lista completa, es decir, todos los
print(lista[::2]) #se obtiene una sublista que contiene todos los elementos de la lista original pero con un paso de 2, es decir, se seleccionan los elementos en las posiciones 0, 2 y 4 de la lista original
print(lista[1:-1]) #se obtiene una sublista que contiene todos los elementos de la lista original excepto el primero y el último
print(lista[1:-1:2]) #se obtiene una sublista que contiene todos los elementos de la lista original excepto el primero y el último, pero con un paso de 2, es decir, se seleccionan los elementos en las posiciones 1 y 3 de la lista original
print(lista[::-1]) #se obtiene una sublista que contiene todos los elementos de la lista original pero en orden inverso, es decir, se seleccionan los elementos desde el último hasta el primero de la lista original
```

- in permite comprobar si un elemento está en una lista

```python
print(6 in lista) # Verificar si el número 6 está presente en la lista, lo que devuelve True si se encuentra y False si no se encuentra
print(10 in lista) # Verificar si el número 10 está presente en la lista
```

- Métodos para agregar elementos: append, que agrega elemento al final de la lista, insert, inserta un elemento en una posición específica y extend que pega los elementos de otro iterable al final. Tambien están + para concatenar y * para repetir la lista.

```python
lista1 = [1, 2, 3] # Crear una lista con algunos elementos
lista2 = [4, 5, 6] # Crear otra lista con algunos elementos
print(lista1 + lista2) # Concatenar las dos listas, lo que devuelve una nueva lista que contiene todos los elementos de ambas listas
lista1.extend(lista2) # modificar la lista lista1 agregando los elementos de la lista lista2 al final de la lista original, lo que modifica la lista original y no devuelve una nueva lista
print(lista1) 
print(lista1 * 3) # Repetir la lista lista1 tres veces, lo que devuelve una nueva lista que contiene los elementos de lista1 repetidos tres veces
lista1.append(7)# Agregar un nuevo elemento al final de la lista lista1, lo que modifica la lista original y no devuelve una nueva lista
print(lista1) 
lista1.insert(1, 10)# Insertar un nuevo elemento en la lista lista1 en la posición 1, lo que modifica la lista original y no devuelve una nueva lista
print(lista1) 

```

- Métodos para elminar elementos: pop, eliminar y retorna el elemento en una posición (si no se indica es el último), remove, elimina la primera aparación de un elemento, clear que vacía una lista.

```python
lista = [1, 2, 3, 4, 5] # Crear una lista con algunos elementos
print(lista) # Imprimir la lista original
lista.pop(3) # Eliminar el elemento en la posición 3 de la lista utilizando el método pop(), si no se indica se elimina el último elemento.
print(lista) # Imprimir la lista después de eliminar el último elemento
lista.remove(2) # Eliminar el primer elemento de la lista que es igual a 2
print(lista) # Imprimir la lista después de eliminar el elemento 2
lista.clear() # Eliminar todos los elementos de la lista utilizando el método clear(), lo que modifica la lista original y no devuelve una nueva lista
print(lista) # Imprimir la lista después de eliminar todos los elementos, lo que devuelve una lista vacía
```

- También hay métodos para ordenar: sort, ordena lista de menor a mayor, y reverse invierte el orden.

```python
lista = [5, 2, 9, 1, 5, 6] # Crear una lista con algunos elementos
print(lista) # Imprimir la lista original
lista.sort() # Ordenar la lista utilizando el método sort(), lo que modifica la lista original y no devuelve una nueva lista
print(lista) # Imprimir la lista después de ordenarla
lista.reverse() # Invertir el orden de los elementos en la lista utilizando el método reverse(), lo que modifica la lista original y no devuelve una nueva lista
print(lista) # Imprimir la lista después de invertir el orden de los elementos
```

- La **list comprenhension** es una forma compacta de crear una nueva lista a partir de un iterable.
- Es más rápida que un ciclo for tradicional y es más limpio de leer
- Su estructura es:
$$[expresión \ for \ elemento \ in \ iterable \ if \ condición]$$

1. **Expresión:** Es el valor u operación que se agregará a la nueva lista, puede usar al elemento del iterable con alguna operación o hasta algún valor constante.
2. **Elemento:** Es una variable temporal (alias) que representa cada elemento del iterable en el ciclo.
3. **Iterable** Es la fuente de los datos (lista, string, range, tupla, etc).
4. **Condición** Es opcional, pero funciona como filtro para decidir si el elememento pasa a la expresión para ser incluido en la lista final.



```python
lista_nombres = ["Samuel", "Ana", "Luis", "Maria", "Carlos"] # Crear una lista de nombres
n_names_i = [] # Crear una lista vacía para almacenar los nombres que contienen la letra "i"s

#forma tradicional sin list comprehension
for name in lista_nombres: # Iterar sobre cada nombre en la lista de nombres 
    if "i" in name: # Verificar si la letra "i" está presente en el nombre actual
        n_names_i.append(name) # Si la letra "i" está presente, agregar el nombre a la lista de nombres con "i"
print(n_names_i) # Imprimir la lista de nombres que contienen la letra "i"
```

```python
lista_nombres = ["samuel", "ana", "luis", "maria", "carlos"] # Crear una lista de nombres
n_names_i = [nombre for nombre in lista_nombres if "i" in nombre] # Se redujo el problema a una linea.
print(n_names_i)
```

```python
print( [n**2 for n in range(5)] ) # Crear una lista de cuadrados de los primeros 5 enteros
print( [n for n in range(11) if n % 5 == 0] ) # Crear una lista de números del 0 al 10 que son divisibles por 5
print(["par" if n % 2 == 0 else "impar" for n in range(11)]) # Crear una lista que indica si los números del 0 al 10 son pares o impares
print([palabra.title() for palabra in lista_nombres]) # Crear una lista de los nombres en la lista de nombres con la primera letra en mayúscula y el resto en minúscula utilizando el método title() para formatear cada nombre.
```
