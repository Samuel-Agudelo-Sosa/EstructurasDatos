# 🔄 Listas Circulares: Teoría y Conceptos

Tal como indica su nombre, las listas circulares son una **variante de implementación** de las listas enlazadas (simples o dobles) donde se elimina el concepto de "final" para crear una estructura cerrada o cíclica.

## Concepto Principal
La característica definitoria de una lista circular es que **el primer nodo (cabeza) y el último nodo (cola) se unen**. 
- Al llegar al final de la lista y pedir el "siguiente", el recorrido comienza de nuevo por el principio.
- **Adiós a los Nulos:** En una lista circular con elementos, **ningún puntero apunta a `None`** (o Nulo).

---

## Variantes

### 1. Lista Circular Simple (Enlazada)
Basada en la lista simple tradicional, donde cada nodo tiene un dato y un puntero al siguiente (`P` o `next`).
* **La modificación:** El puntero *siguiente* del **último nodo** de la lista ya no apunta a nulo, sino que se enlaza de vuelta al **primer nodo** (cabeza).

### 2. Lista Circular Doblemente Enlazada
Basada en la lista doble, donde cada nodo tiene referencias hacia adelante (`Ps`) y hacia atrás (`Pa`).
* **La modificación (Doble conexión):** 1. El puntero *siguiente* del **último nodo** apunta al **primer nodo**.
  2. El puntero *anterior* del **primer nodo** apunta al **último nodo**.

---

## ⚠️ Peculiaridades en las Operaciones

Las operaciones básicas (Insertar, Borrar, Buscar) mantienen la misma lógica de "puenteo" de referencias (auxAnterior, auxPosterior) que en las listas lineales. Sin embargo, al programarlas hay que tener en cuenta dos reglas de oro:

1. **Evitar el bucle infinito:** Como ya no hay un `None` al final, no puedes usar la clásica condición `while actual is not None:` para recorrer la lista. Si lo haces, el programa nunca se detendrá. En su lugar, el bucle debe detenerse cuando el nodo actual o el siguiente vuelva a ser igual a la **cabeza** (`while actual.next != self.head:`).
2. **Mantenimiento del cierre:** Cualquier inserción o borrado que ocurra en los extremos (ya sea cambiar el primer elemento o el último) obliga a reconfigurar el "cierre" del círculo, asegurándose de que la cola y la cabeza siempre queden conectadas.

# Implementación lista circular Simple y Doble

Se añaden las clases que representan a los nodos:

```python
class Nodo: #lista circular simple
    def __init__(self, data, next=None):
        self.data = data
        self.next = next

    def __str__(self): #implementación para el usuario
        return str(self.data)
    
    def __repr__(self): #implementación para el programador
        return f"Nodo({self.data})"
```

```python
class NodoDoblementeEnlazado: #lista circular doblem
    
    def __init__(self, dato, next=None, previous=None):
        self.dato = dato
        self.previous = previous
        self.next = next
    
    def __str__(self):
        return str(self.dato)
    
    def __repr__(self):
        return f"Nodo({self.dato})"
```

Se añade la clase de las estructuras lista circular simple y lista circular doble

```python
class ListaCircular:
    def __init__(self, elementos=None):
        self.head = None 
        self.tail = None 
        if elementos: 
            for item in elementos:
                self.agregar(item)
        
    def agregar(self, data):
        nuevo_nodo = Nodo(data)
        if not self.head:
            self.head = nuevo_nodo
            self.tail = nuevo_nodo 
            self.tail.next = self.head
        else: 
            self.tail.next = nuevo_nodo
            self.tail = nuevo_nodo 
            self.tail.next = self.head
    def __str__(self):
       
        nodos = []
        actual = self.head
        if not actual:
            return "None"
        while True:
            nodos.append(str(actual))
            actual = actual.next
            if actual == self.head:
                break
        return " -> ".join(nodos) + " -> (vuelve al inicio)"
    
    def __repr__(self):
        nodos = []
        actual = self.head
        if not actual:
            return "None"
        while True:
            nodos.append(repr(actual))
            actual = actual.next
            if actual == self.head:
                break
        return f"ListaCircular([{', '.join(nodos)}])"
```

```python
class ListaCircularDoble:

    def __init__(self, elementos=None):
        self.head = None
        self.tail = None
        if elementos:
            for item in elementos:
                self.agregar(item)

    def agregar(self, data):
        nuevo_nodo = NodoDoblementeEnlazado(data)
        if not self.head:
            self.head = nuevo_nodo
            self.tail = nuevo_nodo
            self.tail.next = self.head
            self.head.previous = self.tail

        else:
            self.tail.next = nuevo_nodo
            nuevo_nodo.previous = self.tail
            self.tail = nuevo_nodo
            self.tail.next = self.head
            self.head.previous = self.tail
        
    def __str__(self):
        nodos = []
        actual = self.head
        if not actual:
            return "None"
        while True:
            nodos.append(str(actual))
            actual = actual.next
            if actual == self.head:
                break
        return "(va al final) <- " + " <-> ".join(nodos) + " -> (vuelve al inicio)"
    
    def __repr__(self):
        nodos = []
        actual = self.head
        if not actual:
            return "None"
        while True :
            nodos.append(repr(actual))
            actual = actual.next
            if actual == self.head:
                break
        return f"ListaCircularDoble([{', '.join(nodos)}])"
        

            
```

- Creación de listas circulares:

```python
lista_circular = ListaCircular([1, 2, 3, 4])
print(lista_circular)
lista_circular_doble = ListaCircularDoble([1, 2, 3, 4])
print(lista_circular_doble)

```
