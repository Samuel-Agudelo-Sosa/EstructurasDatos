# Tipo Abstracto de Datos (TAD): Listas

Las **listas** son estructuras de datos lineales consistentes en una secuencia de nodos. Cada nodo permite almacenar datos y contiene referencias (punteros) a los nodos previos o posteriores.

## Características Principales
* **Homogéneas:** Todos sus elementos son del mismo tipo de dato.
* **Dinámicas:** Pueden crecer o decrecer de forma infinita (limitado solo por la memoria), a diferencia de los arrays estáticos.
* **Almacenamiento secuencial:** Los elementos se almacenan uno detrás de otro.

## Clasificación
Las listas se pueden clasificar según dos enfoques fundamentales:
1. **Ordenación:** Listas *ordenadas* y *no ordenadas*.
2. **Forma de enlace:** Listas *simplemente enlazadas* y *doblemente enlazadas*.

## Operaciones Habituales
* `Inicializar()`: Genera una nueva lista vacía (si había datos previos, se suprimen).
* `Insertar(Elemento)`: Inserta un elemento en la lista. (Las listas permiten elementos repetidos).
* `Borrar(Elemento)` / `Borrar(posición)`: Elimina un elemento específico por su valor o por su índice.
* `Obtener(posición)`: Devuelve el elemento que se encuentra en la posición indicada.
* `Localizar(Elemento)`: Devuelve la posición en la que se encuentra un elemento específico.
* `Tamaño()`: Devuelve la cantidad total de elementos en la lista.

> *Nota:* Estas son las operaciones base; casi cualquier otra operación requerida se puede implementar a partir de estas.

# 3.3. Listas enlazadas y doblemente enlazadas

El TAD lista está compuesto por una serie de nodos enlazados mediante punteros. Dependiendo del número de referencias por nodo, pueden ser listas simplemente enlazadas o doblemente enlazadas.

## Listas Enlazadas (Simples)
Es la forma más básica de una lista dinámica. Se puede visualizar como los vagones de un tren, donde cada nodo tiene dos componentes fundamentales:
* **Dato:** El elemento o valor que almacena el nodo (el "vagón").
* **Puntero (P):** La referencia que apunta al siguiente nodo de la secuencia (el "enganche").

**Ejemplo de estado inicial:** `[Tren|P] -> [Avión|P] -> [Coche|P] -> [Moto|P]`
*(Donde "Tren" ocupa el índice 0 y "Moto" el índice 3).*

### Elementos clave para el manejo de la lista
Para operar correctamente sobre una lista, se necesitan referencias externas a los nodos:
1.  **Puntero a la Cabeza (Head):** Es imprescindible. Indica dónde empieza la lista (el primer elemento).
2.  **Puntero a la Cola (Tail):** Es muy recomendable. Apunta al último elemento y permite insertar nuevos datos al final sin tener que recorrer toda la lista.
3.  **Tamaño (Size):** Un atributo opcional pero eficiente para llevar la cuenta de los nodos sin iterar.

---

## Operación: Añadir Elementos

### 1. Añadir al final de la lista
Supongamos que queremos añadir el elemento "Barco" al final de nuestra lista de ejemplo.

* **Estado inicial:** La cabeza es `Tren` y la cola es `Moto`.
* **Transición:**
    1. Se crea el nuevo nodo `Barco`.
    2. Al tener la referencia de la cola (`Moto`), simplemente se enlaza el puntero de `Moto` para que apunte a `Barco`.
    3. Se actualiza el puntero general de la cola para que ahora referencie a `Barco`.
* **Nota:** Si no tuviéramos un puntero a la cola, habría que iterar desde la cabeza nodo por nodo hasta encontrar el que no tiene un "siguiente", y ahí realizar la inserción.

### 2. Insertar en una posición determinada
Supongamos que queremos insertar "Barco" en la posición 2 (actualmente ocupada por "Coche"), desplazando los demás elementos hacia la derecha.

* **Estado inicial:** `Tren(0) -> Avión(1) -> Coche(2) -> Moto(3)`
* **Transición de estados (Algoritmo):**
    1. **Búsqueda:** El algoritmo busca la posición anterior a la deseada (posición 1, nodo `Avión`).
    2. **Resguardo:** Se guarda el nodo que actualmente ocupa la posición 2 (`Coche` y por consiguiente, el resto de la cadena unida a él) en un nodo o puntero auxiliar. Esto rompe temporalmente el enlace original.
    3. **Nuevo enlace 1:** Se modifica el puntero del nodo en la posición 1 (`Avión`) para que apunte al nuevo nodo (`Barco`). *Estado temporal: `Tren -> Avión -> Barco`*.
    4. **Nuevo enlace 2:** Se modifica el puntero del nuevo nodo (`Barco`) para que apunte al nodo que guardamos en el auxiliar (`Coche`).
* **Estado final:** `Tren(0) -> Avión(1) -> Barco(2) -> Coche(3) -> Moto(4)`. La estructura original del resto de los nodos se mantiene intacta.

---

## Operación: Borrar 
La operación de borrado implica eliminar nodos y gestionar sus punteros para no romper la cadena. Fundamentalmente, existen tres escenarios de borrado que requerirán algoritmos ligeramente distintos:
* Borrado del primer nodo (cabeza).
* Borrado del último nodo (cola).
* Borrado de cualquier nodo intermedio.

# 3.4. Operación: Borrar Nodos en Listas Enlazadas

La operación de borrado implica eliminar un nodo de la secuencia asegurando que la cadena no se rompa y los punteros se actualicen correctamente. Existen tres escenarios principales:

## 1. Borrado del último nodo (Cola)
El objetivo es eliminar el nodo que se encuentra al final de la lista.

* **Estado inicial:** `... -> Barco(2) -> Coche(3) -> Moto(4) -> null`
    * La `Cabeza` apunta a `Tren` y la `Cola` apunta a `Moto`.
* **Transición de estados (Algoritmo):**
    1. **Búsqueda del penúltimo:** Se debe recorrer la lista desde el principio buscando un nodo (A) cuyo siguiente nodo (B) apunte a `null`. Una estrategia es usar dos iteradores simultáneos (`penúltimo` y `último`).
    2. En este caso, se detecta que `Coche` apunta a `Moto`, y `Moto` apunta a `null`.
    3. **Desvinculación:** Se modifica el puntero del nodo anterior (`Coche`) para que ahora apunte a `null`.
    4. **Actualización:** Se actualiza el puntero de la `Cola` para que ahora referencie a `Coche`.
* **Estado final:** `... -> Barco(2) -> Coche(3) -> null`. El nodo `Moto` ha sido eliminado estructuralmente.

---

## 2. Borrado del primer nodo (Cabeza)
El objetivo es eliminar el nodo inicial para que el segundo nodo pase a ser la nueva cabeza.

* **Estado inicial:** `Tren(0) -> Avión(1) -> ...`
    * La `Cabeza` apunta a `Tren`.
* **Transición de estados (Algoritmo):**
    1. **Comprobación:** Verificar si existe un nodo después de la cabeza (para evitar errores o saber si la lista quedará vacía).
    2. **Resguardo:** Se crea un nodo auxiliar que guarda la cabeza actual (`Tren`).
    3. **Actualización de Cabeza:** Se modifica el puntero de la `Cabeza` de la lista para que referencie a su nodo siguiente (`Avión`). *Estado temporal: `Cabeza = Avión`*.
    4. **Limpieza:** Se borra/anula el nodo auxiliar que contenía la antigua cabeza (`Tren` = `null`).
* **Estado final:** `Avión(0) -> Barco(1) -> ...`. El nodo `Tren` desaparece y `Avión` es el nuevo primer elemento.

---

## 3. Borrado de cualquier otro nodo (Intermedio)
El objetivo es eliminar un nodo que no es ni el primero ni el último, reconectando la lista a su alrededor.

* **Estado inicial:** `Tren(0) -> Avión(1) -> Barco(2) -> Coche(3) -> Moto(4)`
    * Supongamos que se desea borrar el nodo en la posición 2 (`Barco`).
* **Transición de estados (Algoritmo):**
    1. **Identificación de actores:** Se debe localizar la `cabeza` para iniciar el recorrido.
    2. **Resguardo de nodos clave:** * Se busca el nodo a borrar y se guarda en una variable auxiliar (`auxNodoBorrado` = `Barco`).
       * Se busca el nodo inmediatamente posterior al que se va a borrar y se guarda (`auxSiguiente` = `Coche`).
       * Se identifica el nodo anterior al que se va a borrar (`Avión`).
    3. **Reconexión (El salto):** Se elimina el vínculo entre el nodo anterior (`Avión`) y el nodo a borrar (`Barco`). Se modifica el puntero del nodo anterior (`Avión`) para que apunte *directamente* al nodo siguiente (`Coche`).
    4. **Limpieza:** Se le da valor nulo al nodo extraído (`auxNodoBorrado` = `null`).
* **Estado final:** `Tren(0) -> Avión(1) -> Coche(2) -> Moto(3)`. El nodo `Barco` ha sido extraído y la cadena se ha cerrado uniendo el nodo 1 con el antiguo nodo 3.

# Implementación LIstas enlazadas en Python

Se añade la clase que representa a los nodos de la lista 

```python
class Nodo:
    def __init__(self, data, next=None):
        self.data = data
        self.next = next

    def __str__(self): #implementación para el usuario
        return str(self.data)
    
    def __repr__(self): #implementación para el programador
        return f"Nodo({self.data})"
```

Se añade la clase que contiene un puntero al inicio de la lista y otro al final.

```python
class ListaEnlazada:
    def __init__(self, elementos=None):
        self.head = None #primerNodo
        self.tail = None #último nodo
        if elementos: #convertir de lista de python a enlazada
            for item in elementos:
                self.agregar(item)
        
    def agregar(self, data):
        nuevo_nodo = Nodo(data)
        if not self.head: #si la lista está vacía entonces su primer y último elemento será el nuevo nodo
            self.head = nuevo_nodo
            self.tail = nuevo_nodo 
        else: #si la lista no está vacía entonces: 
            self.tail.next = nuevo_nodo#el siguiente nodo del último nodo será el nuevo nodo
            self.tail = nuevo_nodo #el nuevo nodo se convierte en el último nodo
    def __str__(self):
        #Representación para el usuario de la lista enlazada: 1 -> 2 -> 3 -> None
        nodos = []
        actual = self.head
        while actual:
            nodos.append(str(actual))
            actual = actual.next
        return " -> ".join(nodos) + " -> None"
    
    def __repr__(self):
        #Representación para el programador de la lista enlazada: ListaEnlazada([1, 2, 3])
        nodos = []
        actual = self.head
        while actual:
            nodos.append(repr(actual))
            actual = actual.next
        return f"ListaEnlazada([{', '.join(nodos)}])"
```

Ejemplos de Uso:

```python
#Creación de lista enlazada
l1 = ListaEnlazada()
print(l1) #Lista vacía
print([l1]) # [ListaEnlazada([])]

first_node = Nodo("a")
second_node = Nodo("b")
third_node = Nodo("c")
print(first_node) #a
```

Creación de lista haciendo manualmente los enlaces:

```python
#Conexión de nodos
first_node.next = second_node
second_node.next = third_node
l1.head = first_node
l1.tail = third_node
print(l1) #a -> b -> c -> None
```

Creación de lista sin necesidad de los nodos:

```python
l2 = ListaEnlazada(["a", "b", "c"]) #convertir de lista de python a enlazada, el constructor de la clase se encarga de agregar cada elemento converitiéndolo en un nodo y conectándolos
print(l2) #a -> b -> c -> None
print([l2]) # [ListaEnlazada([Nodo(a), Nodo(b), Nodo(c)])]
print(l2.head.next) # Nodo(b)
print([first_node, second_node, third_node]) # [Nodo(a), Nodo(b), Nodo(c)]
```

- Tamaño lista enlazada procedimiento:

```python
def tamaño(lista):
    acc = 0
    elemento_actual = lista.head
    while elemento_actual is not None:
        acc += 1
        elemento_actual = elemento_actual.next
    return acc

print(tamaño(l1)) #3
```

- Procedimiento para buscar elemento:

```python
def BuscarElemento(lista, x):
    acc = 0
    elemento_actual = lista.head
    while elemento_actual is not None:
        acc += 1
        if elemento_actual.data == x:
            return True
        elemento_actual = elemento_actual.next  
    return False

print(BuscarElemento(l1, "b")) #True
print(BuscarElemento(l1, "d")) #False
```
