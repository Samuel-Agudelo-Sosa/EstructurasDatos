## Pila Dinámica: Concepto, Lógica y Flujo de Nodos

A diferencia de una pila estática, una **pila dinámica** se construye mediante la consecución de nodos, utilizando un sistema idéntico al de las listas enlazadas. Las operaciones y el funcionamiento general son muy similares, pero con la diferencia de que no se dispone de un array para introducir los elementos. 

Todo su funcionamiento se sustenta en el **enlace de nodos**.

---

### Lógica de las Operaciones Principales (con Ejemplo)

Vamos a simular una pila donde guardaremos palabras.

* **1. Creación e Inicialización:** Al crear la pila, se establecen parámetros básicos. El elemento central es el **«tope» o cima**, que nos establece dónde se encuentra la pila. Dado que todavía no se tienen elementos, el tope apunta a `null` (es un nodo vacío).
  > **Ejemplo visual:** > `Tope -> null`

* **2. Insertar el primer valor (Push):** Al insertar el primer valor, el tope pasa a ser un nodo con dicho valor y cuyo elemento «siguiente» sigue apuntando a `null`, pues no hay más valores posibles por debajo.
  > **Ejemplo visual:** Insertamos "Coche".
  > `Tope -> [ "Coche" | siguiente ] -> null`

* **3. Añadir nuevos elementos (Push):** Cada vez que ingresa un nuevo elemento, este se convierte en el nuevo tope. El puntero "siguiente" del nuevo nodo se enlaza al elemento que era el tope anteriormente.
  > **Ejemplo visual:** Insertamos "Moto" y luego "Barco".
  > *Al insertar "Moto":*
  > `Tope -> [ "Moto" | sig ] -> [ "Coche" | sig ] -> null`
  > *Al insertar "Barco":*
  > `Tope -> [ "Barco" | sig ] -> [ "Moto" | sig ] -> [ "Coche" | sig ] -> null`

* **4. Obtener la cima (Peek):** Simplemente es necesario obtener el valor al que está apuntando el tope. 
  > **Ejemplo:** Si hacemos `peek()`, el sistema mira el `Tope` y nos devuelve `"Barco"` (sin borrarlo).

* **5. Desapilar (Pop):** Se encarga de devolver el elemento al que apunta el tope y **actualizar la referencia**. El nuevo tope pasa a ser el nodo al que apuntaba el elemento recién extraído.
  > **Ejemplo visual:** Hacemos `pop()`. El sistema nos devuelve `"Barco"`. 
  > Ahora el `Tope` se mueve al "siguiente" de Barco, que es Moto:
  > `Tope -> [ "Moto" | sig ] -> [ "Coche" | sig ] -> null`

---

### El sentido de los enlaces: ¿Por qué de Derecha a Izquierda?

Como se puede observar en las pilas, a diferencia de las listas, los elementos se van insertando de derecha a izquierda. Esto significa que el primer elemento que entra queda a la derecha, y los nuevos se van apilando a su izquierda, apuntando siempre hacia los más antiguos.

Esto es debido a que es la **única forma** (trabajando con un solo enlace: siguiente) de que se pueda desapilar un elemento y volver al que le sigue usando el único camino posible. Si se hiciera de izquierda a derecha, con un solo enlace, sería imposible.

**Demostración práctica del problema:**

✅ **Cómo funciona actualmente (Derecha a Izquierda / El nuevo apunta al viejo):**
`Tope -> [ "Barco" | sig ] -> [ "Moto" | sig ] -> [ "Coche" | sig ] -> null`
Si sacamos "Barco" (`pop`), el propio nodo de "Barco" tiene una flecha (`sig`) que nos dice "el que estaba debajo de mí era Moto". El Tope se actualiza fácilmente siguiendo esa flecha.

❌ **Por qué es imposible de Izquierda a Derecha (El viejo apunta al nuevo):**
Imaginemos que insertamos como en una cola normal:
`[ "Coche" | sig ] -> [ "Moto" | sig ] -> [ "Barco" | sig ] -> null`
Aquí el `Tope` estaría obligatoriamente en `"Barco"` (el último en entrar).
`[ Coche ] -> [ Moto ] -> [ Barco ] <- Tope`
Si hacemos un `pop` para sacar `"Barco"`, necesitamos que el `Tope` retroceda a `"Moto"`. **¡Pero no podemos!** El nodo `"Barco"` apunta a `null`, no a `"Moto"`. La flecha va de Moto a Barco, pero no podemos viajar en contra de la flecha con un solo enlace. Habríamos perdido la referencia de la pila por completo.

# Implementación Pilas dinámicas en python.

Se añade la clase nodo:

```python
class Node:
    #Clase que representa un elemento individual en la pila, con un valor y un puntero al siguiente nodo.
    def __init__(self, data):
        self.data = data
        self.next = None  # Puntero al siguiente nodo

    def __str__(self):
        return str(self.data)
    
    def __repr__(self):
        return f"Node({self.data})"
```

Se añade la clase LinkedStack:

```python
class LinkedStack:
    
    # La clase LinkedStack implementa una pila utilizando una estructura de nodos enlazados.
    
    def __init__(self):
        # Utilizamos un "nodo centinela"
        # para que siempre exista un tope, simplificando la lógica de inserción.
        self.tope = Node("Tope")
        self.tope.next = None

    def push(self, e):
        # Para agregar un nuevo elemento, creamos un nuevo nodo con el valor dado.
        elemento = Node(e)
        # El nuevo elemento apunta al que antes era el primero
        elemento.next = self.tope.next
        # El centinela ahora apunta al nuevo elemento, dejándolo en la cima
        self.tope.next = elemento

    def empty(self):
        # La pila está vacía si el nodo centinela no apunta a ningún otro nodo.
        return self.tope.next is None
    
    def peek(self):
        # Devuelve el valor de la cima sin modificar la pila.
        if self.empty():
            raise Exception("Empty stack") # Requisito de la rúbrica
        return self.tope.next.data

    def pop(self):
        # Elimina y devuelve el valor de la cima de la pila.
        if self.empty():
            raise Exception("Empty stack")
        
        # Guardamos el valor para retornarlo al final
        valor = self.tope.next.data
        # Actualizamos el puntero: saltamos el primer nodo, enlazando con el siguiente
        self.tope.next = self.tope.next.next
        return valor
    
    def clear(self):
        # Vacía la pila simplemente apuntando el nodo centinela a None, eliminando todas las referencias a los nodos anteriores.
        self.tope.next = None
    
    def __str__(self):
        # Construye una representación legible de la pila, mostrando los elementos desde la cima hasta el fondo.
        result = []
        current = self.tope # Imprime desde el nodo centinela.
        while current is not None:
            result.append(str(current.data))
            current = current.next
        return " -> ".join(result) + " -> None"
    
    def __repr__(self):
        # Proporciona una representación más técnica de la pila, mostrando el nodo que sigue al centinela.
        return f"LinkedStack({self.tope.next})"
```

Ejemplos de uso:

```python

# 1. Inicialización
mi_pila = LinkedStack()
print("1. ¿La pila está vacía al inicio?:", mi_pila.empty())








```

```python
# 2. Apilar elementos (push)
print("\n2. Apilando elementos (10, 20, 30)...")
mi_pila.push(10)
mi_pila.push(20)
mi_pila.push(30)

# Comprobamos la representación __str__
print("Estado de la pila:", mi_pila) 
# Salida esperada: 30 -> 20 -> 10 -> None (nota cómo el 30 quedó en la cima/izquierda)

# Comprobamos la representación __repr__
print("Representación interna:", repr(mi_pila))
```

```python
# 3. Ver la cima (peek)
print("\n3. Elemento en la cima (peek):", mi_pila.peek())
print("¿La pila cambió después de peek?:", mi_pila) # Debería seguir igual
```

```python
# 4. Desapilar un elemento (pop)
elemento_extraido = mi_pila.pop()
print(f"\n4. Elemento extraído con pop: {elemento_extraido}")
print("Estado de la pila tras el pop:", mi_pila)
```

```python
# 5. Comprobar estado actual
print("\n5. ¿La pila está vacía ahora?:", mi_pila.empty())
```

```python


# 6. Vaciar la pila (clear)
print("\n6. Vaciando la pila (clear)...")
mi_pila.clear()
print("Estado de la pila tras clear:", mi_pila)
print("¿Está vacía?:", mi_pila.empty())
```

```python
# 7. Probar las excepciones
print("\n7. Probando las excepciones de pila vacía...")
try:
    mi_pila.pop()
except Exception as e:
    print(f"Excepción capturada correctamente al hacer pop: {e}")

try:
    mi_pila.peek()
except Exception as e:
    print(f"Excepción capturada correctamente al hacer peek: {e}")
```
