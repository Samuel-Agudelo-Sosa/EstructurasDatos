# Listas Doblemente Enlazadas

Las **listas doblemente enlazadas** son una extensión de las listas enlazadas simples. Su principal característica es que cada nodo guarda una referencia (puntero) tanto al nodo que le precede como al que le sigue.

## Estructura de un Nodo
A diferencia de las listas simples, en las listas doblemente enlazadas existen tres componentes clave por nodo:
* **Pa:** Puntero o referencia al nodo **anterior**.
* **Dato:** La información o valor almacenado.
* **Ps:** Puntero o referencia al nodo **siguiente**.

A efectos prácticos, puede visualizarse como dos listas enlazadas simples que contienen los mismos datos, pero en orden inverso.

---

## Ventajas y Desventajas

### Ventajas
* **Recorrido Bidireccional:** Permiten que la lista sea recorrida en ambos sentidos (hacia adelante y hacia atrás).
* **Operaciones de búsqueda más eficientes:** Si necesitas acceder al nodo previo de un elemento dado, puedes hacerlo directamente sin necesidad de recorrer toda la lista desde el principio.

### Desventajas
* **Mayor costo en inserción/eliminación:** Añadir o borrar elementos es computacionalmente "más costoso" porque requiere actualizar un mayor número de enlaces (punteros) en comparación con una lista simple.

---

## Ejemplo de Eficiencia Computacional
Imagina una lista con miles de elementos (ej. nodo en el índice 4568) y necesitas retroceder una sola posición (al índice 4567):
* **En una lista simple (L1):** Debes colocarte en la cabecera (nodo 0) y avanzar 4567 posiciones.
* **En una lista doblemente enlazada (L2):** Simplemente utilizas el puntero **Pa** para retroceder una única posición. Esto ilustra una enorme ventaja en tiempo de ejecución.

---

## Particularidades (Nodos de los Extremos)
Para que la estructura funcione correctamente y no genere errores al recorrerla, los extremos de la lista tienen características especiales:
1. **El primer nodo (Cabeza):** No tiene un nodo anterior. Su puntero **Pa** apunta a un valor nulo (`null`). Esto indica que se ha llegado al principio si se recorre de forma inversa.
2. **El último nodo (Cola):** No tiene un nodo posterior. Su puntero **Ps** apunta a un valor nulo (`null`), indicando el final convencional de la lista.

---

# Operaciones en Listas Doblemente Enlazadas

En las listas doblemente enlazadas, las operaciones de inserción y borrado en el medio de la lista son las más complejas debido a la necesidad de gestionar referencias bidireccionales (hacia atrás y hacia adelante). 

A continuación, se detalla la lógica paso a paso para ambas operaciones.

## 1. Insertar un nodo en el medio

**Objetivo:** Añadir un nuevo elemento (ej. *Barco*) entre dos nodos existentes (ej. *Coche* y *Moto*).

**Paso a paso:**
1. **Identificar nodos adyacentes:** Guardar en una variable auxiliar (`auxAnterior`) el nodo que quedará antes del nuevo elemento (Coche).
2. **Identificar nodo a desplazar:** Guardar en otra variable auxiliar (`auxPosterior`) el nodo que actualmente ocupa la posición deseada y que se desplazará a la derecha (Moto).
3. **Configurar el nuevo nodo:** * Asignar su puntero *anterior* (`Pa`) para que apunte a `auxAnterior`.
   * Asignar su puntero *siguiente* (`Ps`) para que apunte a `auxPosterior`.
4. **Reconfigurar los nodos adyacentes:**
   * Modificar el puntero *siguiente* (`Ps`) de `auxAnterior` para que apunte al nuevo nodo.
   * Modificar el puntero *anterior* (`Pa`) de `auxPosterior` para que apunte al nuevo nodo.

---

## 2. Borrar un nodo del medio

**Objetivo:** Eliminar un elemento específico (ej. *Barco*) que se encuentra entre otros dos nodos (*Coche* y *Moto*).

**Paso a paso:**
1. **Identificar el nodo previo:** Guardar en `auxAnterior` el nodo que precede al elemento que se va a borrar (Coche).
2. **Identificar el nodo posterior:** Guardar en `auxPosterior` el nodo que va después del elemento a borrar (Moto).
3. **Aislar el nodo a borrar:** Guardar el nodo objetivo en una variable `auxNodoBorrado`.
4. **Puenteo de enlaces (Desreferenciación):**
   * Cambiar el puntero *siguiente* (`Ps`) de `auxAnterior` para que apunte directamente a `auxPosterior`.
   * Cambiar el puntero *anterior* (`Pa`) de `auxPosterior` para que apunte directamente a `auxAnterior`.
5. **Eliminación final:** Establecer el nodo `auxNodoBorrado` (o sus referencias internas) como **Nulo/None** para borrar definitivamente sus conexiones de salida, permitiendo que el recolector de basura libere la memoria.

# Implementación de listas doblemente enlazadas en python

Se añade la clase que representa a los nodos:

```python
class NodoDoblementeEnlazado:
    
    def __init__(self, dato, next=None, previous=None):
        self.dato = dato
        self.previous = previous
        self.next = next
    
    def __str__(self):
        return str(self.dato)
    
    def __repr__(self):
        return f"Nodo({self.dato})"
```

Se añade la clase que contiene un puntero al inicio de la lista y otro al final:

```python
class Nodo:
    def __init__(self, data, next=None):
        self.data = data
        self.next = next

    def __str__(self): #implementación para el usuario
        return str(self.data)
    
    def __repr__(self): #implementación para el programador
        return f"Nodo({self.data})"

class ListaDoblementeEnlazada:

    def __init__(self, elementos=None):
        self.head = None
        self.tail = None
        if elementos:
            for item in elementos:
                self.agregar(item)

    def agregar(self, data):
        nuevo_nodo = NodoDoblementeEnlazado(data)
        if not self.head:
            self.head = nuevo_nodo
            self.tail = nuevo_nodo
           #nuevo_nodo.previous = None #no es necesario porque el constructor ya lo establece como None
            #nuevo_nodo.next = None #no es necesario porque el constructor ya lo establece como None
        else:
            self.tail.next = nuevo_nodo
            nuevo_nodo.previous = self.tail
            self.tail = nuevo_nodo
            #nuevo_nodo.next = None #no es necesario porque el constructor ya lo establece como None

    def __str__(self):
        nodos = []
        actual = self.head
        if not actual:
            return "None"
        while actual:
            nodos.append(str(actual))
            actual = actual.next
        return "None <- " + " <-> ".join(nodos) + " -> None"
    
    def __repr__(self):
        nodos = []
        actual = self.head
        while actual:
            nodos.append(repr(actual))
            actual = actual.next
        return f"ListasDoblementeEnlazadas([{', '.join(nodos)}])"
        

            
```

Ejemplos de uso:

```python
#Creación de lista enlazada
l1 = ListaDoblementeEnlazada()
print(l1) #Lista vacía
print([l1]) # [ListasDoblementeEnlazadas([])]

first_node = NodoDoblementeEnlazado("a")
second_node = NodoDoblementeEnlazado("b")
third_node = NodoDoblementeEnlazado("c")
print(first_node) #a
```

Creación de las listas haciendo manualmente los enlaces:

```python
#Conexión de nodos
first_node.next = second_node
second_node.next = third_node
second_node.previous = first_node
third_node.previous = second_node
#first_node y third_node no necesitan previous ni next respectivamente porque el constructor ya los establece como None
l1.head = first_node
l1.tail = third_node
print(l1) # None <- a <-> b <-> c -> None
```

Creación de las listas sin necesidad de nodos:

```python
l2 = ListaDoblementeEnlazada(["a", "b", "c"]) #convertir de lista de python a enlazada, el constructor de la clase se encarga de agregar cada elemento converitiéndolo en un nodo y conectándolos
print(l2) # None <- a <-> b <-> c -> None
print([l2]) # [ListasDoblementeEnlazadas([Nodo(a), Nodo(b), Nodo(c)])]
print(l2.head.next) # Nodo(b)
print(l2.head.next.previous) # Nodo(a)
print([first_node, second_node, third_node]) # [Nodo(a), Nodo(b), Nodo(c)]
```

- Orden inverso de la lista:

```python
def ordenInverso(l):
    elemento_actual = l.tail
    l.head, l.tail = l.tail, l.head
    while elemento_actual:
            elemento_antiguo = elemento_actual.previous
            elemento_actual.previous, elemento_actual.next = elemento_actual.next, elemento_antiguo
            elemento_actual = elemento_antiguo  

print(l1) # None <- a <-> b <-> c -> None
ordenInverso(l1)
print(l1) # None <- c <-> b <-> a -> None
ordenInverso(l1) #volvemos a la normalidad

```
