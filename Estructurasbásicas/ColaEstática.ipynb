# TAD Cola (Estructura de datos FIFO)

El Tipo Abstracto de Datos (TAD) cola es una estructura de datos basada en el principio **FIFO** (*First In, First Out*). Esto significa que **el primer elemento que entra es el primero en salir**, replicando el comportamiento de las colas que encontramos en nuestra vida diaria, como en la caja de un supermercado o en la taquilla del cine.

## Concepto principal

La característica fundamental de una cola es que **no permite el acceso a elementos intermedios**. Se debe respetar estrictamente el orden de llegada y salida de los datos:
* Solo es accesible (y eliminable) el primer elemento de la cola, el cual se encuentra en la **cabeza** (frente) de la misma.
* La **adición** de nuevos elementos se realiza única y exclusivamente al final de la estructura, es decir, tras el último elemento ingresado.

## Tipos de implementación

Dependiendo de cómo se gestione la memoria y el almacenamiento de los elementos, las colas pueden implementarse de dos maneras principales:

* **Colas Estáticas:** Su implementación viene dada por el uso de una estructura de datos auxiliar de tamaño fijo (generalmente un *array* o arreglo). Requieren que se establezca un tamaño máximo previamente. Si la cola se llena, no admitirá más elementos.
* **Colas Dinámicas:** No tienen un tamaño máximo predefinido rígido (están limitadas solo por la memoria del sistema). La reserva de memoria para almacenar los elementos se realiza sobre la marcha a medida que estos van ingresando a la estructura.

## Operaciones básicas

Independientemente de su implementación, un TAD cola debe soportar una serie de operaciones fundamentales para su manipulación:

* **Inicializar():** Corresponde al proceso de preparación y creación de la cola. En el caso de las colas estáticas, es el momento donde se realiza la reserva de memoria (creación del *array*) con el tamaño máximo deseado.
* **Encolar(Elemento):** Es la acción de añadir un nuevo elemento. Este siempre se inserta al final de la cola. Suele devolver un indicador (como un valor booleano o una excepción) si ocurre un error, por ejemplo, si se intenta encolar en una cola estática que ya está llena.
* **Desencolar():** Obtiene, devuelve y elimina el elemento que está más adelante en la cola (en la posición de la cabeza), respetando la estrategia FIFO. Si la cola está vacía, debe notificar un error o lanzar una excepción.
* **Primero() / Cima / Tope:** Devuelve el valor del elemento que ocupa la primera posición en la cola (el próximo a salir) **sin eliminarlo** de la estructura.
* **Vacía():** Es una comprobación que devuelve un valor lógico indicando si la cola carece de elementos en ese momento.

## Lógica interna de control: Enfoque estático

Cuando se implementa una cola de forma estática utilizando un *array* tradicional (no circular), la lógica de control interno se maneja habitualmente mediante un puntero o índice (a menudo llamado `ultimoElemento`):

1.  **Estado Inicial:** Se crea un *array* de tamaño fijo (ej. 8 posiciones). El puntero `ultimoElemento` se inicializa en `0`, indicando la primera posición libre.
2.  **Proceso de Encolar:** El nuevo dato se guarda en la posición que marca el puntero `ultimoElemento`. Inmediatamente después, el puntero se incrementa en `1` para apuntar a la siguiente casilla libre.
3.  **Proceso de Desencolar:** Es la operación más compleja en este enfoque. 
    * Se extrae el valor ubicado en la posición `0` del *array* (la cabeza de la cola).
    * Al quedar la posición `0` libre, es obligatorio realizar una **reordenación**: se deben desplazar todos los elementos restantes una posición hacia la izquierda.
    * Finalmente, el puntero `ultimoElemento` retrocede una posición (`-1`) para reflejar correctamente el nuevo espacio disponible al final.

*(Nota: Para evitar el costoso desplazamiento de elementos hacia la izquierda, en implementaciones más avanzadas se suele utilizar el concepto de **Cola Circular**, donde se emplean dos punteros —uno para el principio y otro para el final— permitiendo aprovechar los espacios vacíos al principio del array simulando una estructura cerrada).*

## Casos de uso y aplicaciones

Las colas son estructuras de datos esenciales en la informática y la programación debido a su naturaleza de procesamiento secuencial justo. Algunas de sus aplicaciones más comunes incluyen:

* **Gestión de tareas en Sistemas Operativos:** Control de los procesos que están listos para usar la CPU (el primer proceso en llegar es el primero en ser atendido por el procesador).
* **Spools de impresión (Colas de impresión):** Los documentos enviados a una impresora se almacenan en una cola; se imprimen estrictamente en el orden en que fueron solicitados.
* **Servidores web y manejo de peticiones:** Los servidores encolan las solicitudes entrantes de los clientes (usuarios) para procesarlas de manera ordenada sin colapsar.
* **Algoritmos de búsqueda:** Son la base de algoritmos fundamentales como la Búsqueda en Anchura (BFS - *Breadth-First Search*) utilizada para recorrer grafos o árboles.
* **Sistemas de atención al cliente:** Software de *Call Centers* o tickets de soporte que asignan el cliente que lleva más tiempo esperando al primer agente disponible.

# Implementación de las colas estáticas en python

Se añade la clase cola:

```python
class Cola:
    def __init__(self, tamaño):
        self.tamaño = tamaño
        self.ultimo = 0
        self.inicializar()
    def inicializar(self):
        self.cola = [0] * self.tamaño
    
    def es_vacia(self):
        return self.ultimo == 0
    
    def encolar(self, dato):
        if self.ultimo < self.tamaño:
            self.cola[self.ultimo] = dato
            self.ultimo += 1
        else:
            print("error")    
    def desencolar(self):
        if self.es_vacia():
            print("error")
            return
        else:
            self.ultimo -= 1
            return self.cola[0]
```

Se añade la clase cola circular que es más eficiente en el proceso de desencolado:

```python
class ColaEstatica:
    def __init__(self, capacidad):
        self.capacidad = capacidad
        self.array = [None] * capacidad  # Creamos el array de tamaño fijo
        self.frente = 0                  # Índice del primer elemento
        self.final = 0                   # Índice donde se insertará el próximo elemento
        self.cantidad_elementos = 0      # Para llevar el control de si está llena o vacía

    def es_vacia(self):
        return self.cantidad_elementos == 0

    def es_llena(self):
        return self.cantidad_elementos == self.capacidad

    def encolar(self, dato):
        if self.es_llena():
            raise Exception("La cola estática está llena")
        
        # Insertamos el dato en la posición 'final'
        self.array[self.final] = dato
        
        # Movemos el puntero 'final' circularmente
        self.final = (self.final + 1) % self.capacidad
        
        self.cantidad_elementos += 1

    def desencolar(self):
        if self.es_vacia():
            raise Exception("La cola estática está vacía")
        
        # Extraemos el dato de la posición 'frente'
        dato = self.array[self.frente]
        self.array[self.frente] = None # Opcional: limpiar la referencia
        
        # Movemos el puntero 'frente' circularmente
        self.frente = (self.frente + 1) % self.capacidad
        
        self.cantidad_elementos -= 1
        return dato

    def __str__(self):
        # Representación visual simple
        return f"Array Interno: {self.array} | Frente: {self.frente} | Final: {self.final}"
    def __repr__(self):
        return self.__str__()
```

Ejemplos de uso:

```python
fila = Cola(3)
print(fila)
fila.encolar(1)
fila.encolar(2)
fila.encolar(3)
fila.encolar(4)
print(fila)
print(fila.desencolar())
print(fila)
```

```python

# --- Pruebas de la Cola Circular ---
mi_cola_estatica = ColaEstatica(3)

mi_cola_estatica.encolar("A")
mi_cola_estatica.encolar("B")
mi_cola_estatica.encolar("C")
print(mi_cola_estatica) 
# Array Interno: ['A', 'B', 'C'] | Frente: 0 | Final: 0 (da la vuelta)

print(f"Desencolo: {mi_cola_estatica.desencolar()}") # Sale la 'A'
print(mi_cola_estatica) 
# Array Interno: [None, 'B', 'C'] | Frente: 1 | Final: 0

# Como desencolamos uno, hay espacio para encolar otro, ¡y dará la vuelta!
mi_cola_estatica.encolar("D")
print(mi_cola_estatica) 


# Array Interno: ['D', 'B', 'C'] | Frente: 1 | Final: 1

mi_cola_estatica.desencolar() # Sale la 'B'
print(mi_cola_estatica)
# Array Interno: ['D', None, 'C'] | Frente: 2 | Final: 1

mi_cola_estatica.desencolar() # Sale la 'C'
print(mi_cola_estatica)
# Array Interno: ['D', None, None] | Frente: 0 | Final: 1

mi_cola_estatica.encolar("E")
mi_cola_estatica.encolar("F")
print(mi_cola_estatica)
# Array Interno: ['D', 'E', 'F'] | Frente: 0 | Final: 1
```
