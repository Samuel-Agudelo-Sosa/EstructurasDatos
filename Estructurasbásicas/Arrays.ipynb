# Fundamentos de Arrays: Memoria y Rendimiento

### 1. Mecanismo de Reserva
Cuando se declara un array (especialmente en lenguajes de bajo nivel como C o C++), el sistema operativo gestiona la memoria siguiendo tres principios:
* **Cálculo de espacio:** Se multiplica la cantidad de elementos por el peso en bytes del tipo de dato. (Ejemplo: 10 enteros $\times$ 4 bytes (sabiendo que el int mide 4 bytes) = 40 bytes en total).
* **Dirección base:** El sistema ubica un bloque contiguo libre de ese tamaño exacto y asigna la dirección del primer byte (índice 0) como punto de partida.
* **Inmutabilidad:** El bloque asignado es de tamaño fijo. No se puede "estirar" posteriormente porque los bytes contiguos podrían estar ocupados por otros procesos o variables.

### 2. Eficiencia y Velocidad (Por qué exigen el mismo tipo de dato)
Los arrays puros logran un acceso instantáneo a cualquier elemento en tiempo constante, o **O(1)**. Esto se debe a que la memoria es contigua y todos los datos miden lo mismo, permitiendo usar una fórmula de salto exacto:

$$\text{Dirección del elemento} = \text{Dirección base} + (\text{índice} \times \text{tamaño del tipo})$$

> **Nota:** Si un array permitiera tipos de datos mezclados (con distintos tamaños en bytes), la fórmula fallaría y la computadora tendría que leer el array linealmente para calcular las posiciones, destruyendo la velocidad.

### 3. Estático vs. Dinámico

| Concepto | Estático | Dinámico |
| :--- | :--- | :--- |
| **Memoria** | Reservada al compilar. Tamaño fijo inamovible. | Reservada en tiempo de ejecución (Heap). |
| **Tipado** | Definido rígidamente en el código (C++, Java). | Descubierto al ejecutar el código (Python, JS). |
| **Estructura** | El array nace y muere con el mismo tamaño. | Arrays como `List` o `ArrayList`. Tienen la "ilusión" de crecer creando un array nuevo más grande, copiando los datos y borrando el viejo. |

### 4. Contexto: Python vs. NumPy
* **Listas de Python (`list`):** Permiten distintos tipos de datos porque, bajo el capó, son **arrays de punteros**. Guardan direcciones de memoria de tamaño fijo que apuntan a donde están los datos reales. Este "doble salto" (buscar la dirección y luego el dato) las hace lentas.
* **Arrays de NumPy (`np.array`):** Restringen los datos a un **único tipo** (como en C). Esto elimina los punteros, guarda los datos de forma contigua en memoria y permite cálculos vectorizados a nivel de procesador, volviéndolos increíblemente rápidos.

# Dimensiones en los Arrays

### 1. Arrays Unidimensionales (1D)
Conocidos comúnmente como **vectores**. Son la estructura matemática y computacional más básica, donde los elementos se organizan en una única línea continua. 
* **Estructura:** Un solo eje (en NumPy conocido como `axis 0`).
* **Acceso:** Se necesita un único índice para ubicar un elemento, representado matemáticamente como $A[i]$.
* **Casos de uso:** Secuencias de datos simples, como una serie de temperaturas diarias o las notas de un estudiante.



````python
import array

array1 = array.array('i', [1, 2, 3, 4, 5]) # Crea un array de enteros con los elementos 1, 2, 3, 4 y 5 y de tipo 'i' (entero con signo)
print(array1) # Imprime el contenido del array
print(type(array1)) # Imprime el tipo de dato del array
print(array1[0]) # Imprime el primer elemento del array (1)
````

### 2. Arrays Bidimensionales (2D)
Conocidos habitualmente como **matrices**. Organizan los datos en un formato tabular de filas y columnas, similar a una hoja de cálculo de Excel.
* **Estructura:** Dos ejes (`axis 0` para las filas y `axis 1` para las columnas).
* **Acceso:** Requieren dos índices para encontrar un dato, expresados como $A[i, j]$ (donde $i$ es la fila y $j$ es la columna).
* **Casos de uso:** Imágenes en escala de grises (una cuadrícula de píxeles), tablas de bases de datos o tableros de juegos (como el ajedrez).

````python
row1 = ["A", "B", "C"]
row2 = ["D", "E", "F"]
row3 = ["G", "H", "I"]

matrix = [row1, row2, row3] # Crea una matriz (lista de listas) con las filas definidas anteriormente
print(matrix) # Imprime la matriz completa
````

### 1. El concepto: Referencias y el problema de `M = [[0]*10]*10`

En Python, las variables no "contienen" los objetos, sino que son **punteros o referencias** (etiquetas) que apuntan a una ubicación en la memoria donde vive el objeto. 

* **El caso del Error:** Cuando hacemos `M = [[0]*10] * 10`, Python evalúa primero el elemento interno `[0]*10` creando **una única lista** en memoria. Luego, el multiplicador externo `* 10` le dice a Python: "crea una nueva lista y pon dentro 10 referencias apuntando exactamente a esa **misma** lista original". 
* **El caso del Objetivo:** Lo que realmente queremos es una lista principal donde cada elemento apunte a una lista en memoria **totalmente independiente y distinta**.

---

### 2. ¿Qué pasa cuando modificamos elementos en el "Caso del error"?

**A. Modificar un elemento específico: `M[0][1] = 1`**
Esto hace que aparezca un `1` en toda esa columna. 
* **Por qué ocurre:** Le estamos diciendo a Python: "Ve a la lista a la que apunta `M[0]`, y cambia su índice `1` por un `1`". Python obedece y modifica esa **única lista compartida** en memoria. Como `M[1]`, `M[2]`, etc., son simples espejos que apuntan a esa misma lista, todos reflejan el cambio simultáneamente.

**B. Modificar la fila completa: `M[0] = [1, 1, 1...]`**
Aquí **no pasa nada malo** con el resto de las filas.
* **Por qué ocurre:** Al usar `=`, no estamos modificando la lista compartida. Estamos **reasignando** el puntero. Le decimos a Python: "A partir de ahora, el índice `0` de `M` ya no apuntará a esa lista compartida; apuntará a esta *nueva* lista". El resto de las filas (`M[1]` a `M[9]`) seguirán apuntando a la vieja lista de ceros.

---



````python
fila1 = [0]*3 # Crea una lista llamada fila1 con 3 elementos, todos inicializados a 0 esto crea una copia de la lista fila1
matriz = [fila1]*3 # Crea una matriz (lista de listas) con 3 filas, cada una es una referencia a la misma lista fila1
#observamos que el * crea referencias a la misma lista fila1, por lo que cualquier cambio en una de las filas afectará a todas las filas, ya que todas son referencias a la misma lista fila1
print(matriz)
matriz[0][0] = 1 # Cambia el valor de la primera celda de todas las filas a 1, ya que todas las filas son referencias a la misma lista fila1
print(matriz)
fila1[1] = 2 # Cambia el valor de la primera celda de todas las filas a 2, ya que todas las filas son referencias a la misma lista fila1
print(fila1)
print(matriz) # Imprime la matriz después de los cambios, mostrando que todas las filas han sido modificadas debido a que son referencias a la misma lista fila1
matriz[0] = [3, 4, 5] # Cambia la primera fila de la matriz a una nueva lista [3, 4, 5], lo que no afecta a las otras filas ya que ahora la primera fila es una referencia a una nueva lista
print(matriz) # Imprime la matriz después de cambiar la primera fila, mostrando que solo
````

````python
matriz = [[0]*3]*3 # Crea una matriz de 3 filas y 3 columnas, pero todas las filas son referencias a la misma lista de ceros
#qué pasa si quiero cambiar una columna? 
for i in range(len(matriz)):
    matriz[i][0] = i+1
print(matriz)  # todas las filas tienen el 3 en la primera columna, porque fue el último valor asignado a esa posición en la lista compartida por todas las filas.
matriz2 = [[0 for x in range(3)] for y in range(3)] # Crea una matriz de 3 filas y 3 columnas, donde cada fila es una lista independiente de ceros
for i in range(len(matriz2)):
    matriz2[i][0] = i+1
print(matriz2) # Imprime la matriz después de cambiar la primera columna, mostrando que cada fila tiene un valor diferente en la primera columna, ya que cada fila es una lista independiente.
````

### 5. Arrays N-dimensionales: Expandiendo el concepto

Una vez dominado el concepto de matriz bidimensional (2D), dar el salto a **matrices multidimensionales** es simplemente agregar más "capas" de referencias. 

La clave para entenderlo es mantener la misma lógica de punteros que vimos antes: dentro de cada posición de un array n-dimensional, no hay un dato final, sino **una referencia (o vector) que apunta a otro array** en la siguiente dimensión. 

Visualmente lo podemos imaginar así:
* **1D (Vector):** Una lista simple de elementos (ej. una línea).
* **2D (Matriz):** Una lista donde cada elemento apunta a un Vector 1D (ej. una cuadrícula o tabla).
* **3D (Cubo):** Una lista donde cada elemento apunta a una Matriz 2D (ej. un cubo, o una caja llena de cuadrículas).
* **N-D:** Esto se puede extender tantas veces como necesites. Cada dimensión adicional es simplemente un array que contiene punteros hacia arrays de la dimensión inferior.

---

#### Ejemplo práctico: Creando un Array 3D

Si aplicamos la forma "Pythónica" (List Comprehension) para asegurarnos de crear objetos independientes en memoria (y evitar el problema de modificar todo a la vez), así crearíamos un "cubo" o matriz tridimensional de 3x3x3 inicializado en ceros:



````python


cubo_3d = [[[0 for _ in range(3)] for _ in range(3)] for _ in range(3)]
print(cubo_3d) # Imprime el cubo 3D de ceros

# Si queremos modificar un único "cubículo" específico:
# Capa 0, Fila 1, Columna 2
cubo_3d[0][1][2] = 5

# Como usamos List Comprehension, solo ese cubículo cambiará, 
# demostrando que nuestras referencias son independientes.
````
