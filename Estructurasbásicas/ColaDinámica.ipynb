# Cola dinámica

A diferencia de la cola estática, la cola dinámica **se construye mediante la consecución de nodos**, al igual que las listas enlazadas. Las operaciones son iguales y el funcionamiento es muy similar, con la diferencia de que **no se dispone de un array** en el que se van introduciendo los elementos.

En primer lugar, la creación de una cola dinámica establece una serie de parámetros básicos. La principal ventaja de las colas respecto a las pilas a la hora de ser implementadas de forma dinámica es la similitud entre estas y las listas, a diferencia de con las pilas, donde **el hecho de que la inserción se realice en sentido contrario** a veces dificulta comprender correctamente cómo funcionan.

## Referencias Básicas
Cuando se crea una cola dinámica se debe partir de dos referencias básicas para controlar qué hay en cada momento en cada posición de la cola:

* **primerElemento:** hace referencia al elemento que va a salir cuando se haga un desencole.
* **ultimoElemento:** hace referencia a dónde se va a insertar un nuevo nodo cuando se encole.

En un primer momento, ambas referencias apuntan al mismo sitio (`None`), ya que no se tienen datos. 

## Operaciones
* **Encolar:** En el momento en el que se inserte un elemento, ambos apuntarán a dicho elemento. A partir de ese momento, al insertar más datos, se irá moviendo el puntero `ultimoElemento` hacia el último nodo insertado.
* **Obtener el primero:** Simplemente es necesario obtener el valor al que está apuntando el puntero `primerElemento`.
* **Desencolar:** Se encarga de devolver el elemento al que apunta el puntero `primerElemento` y actualizar la referencia para que `primerElemento` apunte al siguiente.

# Implementación de las colas dinámicas en Python:

Se añaden las listas enlazadas, ya que permiten que el proceso de desencolar sea O(1), incluso con un método de borrar el primer elemento:

```python
class Nodo:
    def __init__(self, data, next=None):
        self.data = data
        self.next = next

    def __str__(self): #implementación para el usuario
        return str(self.data)
    
    def __repr__(self): #implementación para el programador
        return f"Nodo({self.data})"

class ListaEnlazada:
    def __init__(self, elementos=None):
        self.head = None #primerNodo
        self.tail = None #último nodo
        if elementos: #convertir de lista de python a enlazada
            for item in elementos:
                self.agregar(item)
        
    def agregar(self, data):
        nuevo_nodo = Nodo(data)
        if not self.head: #si la lista está vacía entonces su primer y último elemento será el nuevo nodo
            self.head = nuevo_nodo
            self.tail = nuevo_nodo 
        else: #si la lista no está vacía entonces: 
            self.tail.next = nuevo_nodo#el siguiente nodo del último nodo será el nuevo nodo
            self.tail = nuevo_nodo #el nuevo nodo se convierte en el último nodo

    def borrar_primer_nodo(self):
        if not self.head:
            return None
        dato = self.head.data
        self.head = self.head.next
        if self.head is None: #si la lista queda vacía, el último nodo también debe ser None
            self.tail = None
        return dato
    def __str__(self):
        #Representación para el usuario de la lista enlazada: 1 -> 2 -> 3 -> None
        nodos = []
        actual = self.head
        while actual:
            nodos.append(str(actual))
            actual = actual.next
        return " -> ".join(nodos) + " -> None"
    
    def __repr__(self):
        #Representación para el programador de la lista enlazada: ListaEnlazada([1, 2, 3])
        nodos = []
        actual = self.head
        while actual:
            nodos.append(repr(actual))
            actual = actual.next
        return f"ListaEnlazada([{', '.join(nodos)}])"
```

Se añade la clase cola:

```python
class Cola:
    def __init__(self):
        self.inicializar()
    
    def inicializar(self):
        self.cola = ListaEnlazada()

    def es_vacia(self):
        return self.cola.head is None
    
    def encolar(self, dato):
        self.cola.agregar(dato)


    def desencolar(self):
        if self.es_vacia():
            print("error")
            return
        else:
            return self.cola.borrar_primer_nodo()
    
    def __str__(self):
        return str(self.cola)
    
    def __repr__(self):
        return repr(self.cola)
    
        
```
