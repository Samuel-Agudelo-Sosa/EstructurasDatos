# Tuplas

- **Inmutables:** No se permiten adiciones, reemplazos, eliminaciones ni borrados.
- **Heterogeneas:** Los elementos pueden ser de cualquier tipo (incluso tuplas o listas).
- **Ordenados:** Se accede a sus elementos mediante índices.

```python
tupla = (1, 2) # las tuplas se definen utiliznado paréntesis
print(tupla)
```

- También cuentan con un constructor de la clase tuple que espera un iterador.

```python
tupla1 = tuple([1, 2, 3]) # también se pueden crear tuplas a partir de listas utilizando la función tuple()
print(tupla)

tupla2 = tuple("Samuel hola") # también se pueden crear tuplas a partir de cadenas de texto utilizando la función tuple()
print(tupla2)
```

- Para considerar una expresión como una tupla no basta con estar entre paréntesis, las comas o el constructor tuple son claves

```python
print( (10)) # Esto no crea una tupla, sino que simplemente evalúa el número 10. 
print((10,))#Para crear una tupla con un solo elemento, se debe incluir una coma después del elemento, como (10,) o usar constructor tuple([10]).
print(("hola")) # Esto no crea una tupla, sino que simplemente evalúa la cadena "hola".
print(([1, 2, 3])) # Esto no crea una tupla, sino que simplemente evalúa la lista [1, 2, 3].
```

- Las tuplas también tienen el método len, y se pueden acceder a sus elementos mediante índices.

```python
tupla = (1, 2, 3, 4, 5, 6, (2,3)) # Crear una tupla con varios elementos
print(tupla[0]) # Acceder al primer elemento de la tupla
print(tupla[-1]) # Acceder al último elemento de la tupla lo mismo que tupla[len(tupla)-1]
print(tupla[6][0]) # Acceder al primer elemento de la tupla que se encuentra en la posición 6 de la tupla, lo que devuelve el número 2
```

- También se pueden hacer particiones o slices y se pueden concatenr dos tuplas con + y repetir con *. Además de in y not in para comprobar existencia en la tupla

```python
tupla1 = (1, 2, 3, 4, 5, 6, (2,3)) # Crear una tupla con varios elementos
tupla2 = (7, 8, 9) # Crear otra tupla con varios elementos
print(tupla[1:4]) # Acceder a una subtupla desde el índice 1 hasta el índice 3, es decir, los elementos en las posiciones 1, 2 y 3 de la tupla original
print(tupla1 + tupla2) # Concatenar las dos tuplas, lo que devuelve una nueva tupla que contiene todos los elementos de ambas tuplas
print(tupla2 * 3) # Repetir la tupla tupla2 tres veces, lo que devuelve una nueva tupla que contiene los elementos de tupla2 repetidos tres veces
print(2 in tupla1) # Verificar si el número 2 está presente en la tupla, lo que devuelve True si se encuentra y False si no se encuentra
print(10 not in tupla1) # Verificar si el número 10 está presente en la
```

- Los operadores <, <=, >, >=, ==, != realizan comparaciones entre tuplas 

```python
tupla1 = (1, 100) 
tupla2 = (3, 4)
print(tupla1 == tupla2) # Comparar dos tuplas para verificar si son iguales, lo que devuelve False en este caso ya que las tuplas contienen elementos diferentes
print(tupla1 != tupla2) # Comparar dos tuplas para verificar si son diferentes, lo que devuelve True en este caso ya que las tuplas contienen elementos diferentes
print(tupla1 < tupla2) #devuelve True en este caso ya que el primer elemento de tupla1 (1) es menor que el primer elemento de tupla2 (3)
print(tupla1 <= tupla2) #devuelve True en este caso ya que el primer elemento de tupla1 (1) es menor o igual que el primer elemento de tupla2 (3)
print(tupla1 > tupla2) #devuelve False en este caso ya que el primer elemento de tupla1 (1) no es mayor que el primer elemento de tupla2 (3)
print(tupla1 >= tupla2) #devuelve False en este caso ya que el primer elemento de tupla1 (1) no es mayor o igual que el primer elemento de tupla2 (3)

```
