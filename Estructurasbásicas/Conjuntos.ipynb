 # Conjuntos
 
- **mutables:** sí se pueden modificar
- **No ordenado:** es una colección no ordenada
- **Sin elementos repetidos:** por definición incluso un conjunto no tiene elementos repetidos

```python

set = {1, 2, 3, 4, 5, 5, 6}  # Creando un conjunto con elementos repetidos
print(set)  # Imprime el conjunto, mostrando solo elementos únicos
conjunto_vacio = set()  # Creando un conjunto vacío, los conjuntos se crean o con llaves {} o con la función set()
print(conjunto_vacio)  # Imprime el conjunto vacío
conjunto_lista = set([1, 2, 3, 4, 5])  # Creando un conjunto a partir de una lista
print(conjunto_lista)  # Imprime el conjunto creado a partir de la lista
```

- Los operadores aritméticos, de comparación y las keywords in y not in permiten hacer las operaciones clásicas entre conjuntos:

```python
# vamos a ejemplificar entre los conjuntos A y B las 12 operaciones que hay

A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

# Unión: A ∪ B
union = A.union(B) # es lo mismo que A | B
print("Unión (A ∪ B):", union)

# Intersección: A ∩ B
interseccion = A.intersection(B) # es lo mismo que A & B
print("Intersección (A ∩ B):", interseccion)

# Diferencia: A - B
diferencia = A.difference(B) # es lo mismo que A - B
print("Diferencia (A - B):", diferencia)

# Diferencia simétrica: A Δ B
diferencia_simetrica = A.symmetric_difference(B) # es lo mismo que A ^ B
print("Diferencia Simétrica (A Δ B):", diferencia_simetrica)


print( "2 pertenece a A:", 2 in A) # Pertenece a A
print( "6 not pertenece a A:", 6 not in A) # No pertenece a A

print( A == B) # A es igual a B
print( A != B) # A no es igual a B

print( A.issubset(B)) # A es subconjunto de B también se puede usar A <= B, y subconjunto propio A < B
print( A.issuperset(B)) # A es superconjunto de B, también se puede usar A >= B, y superconjunto propio A > B
```

- También se pueden recorrer los elementos de un conjunto con algún for o incluso con list comprehension

```python
#generar conjunto de números y buscar los pares con for y con list comprension

numeros = set(range(1, 21)) # Genera un conjunto de números del 1 al 20

# Buscar los números pares usando un bucle for
pares_for = set()
for numero in numeros:
    if numero % 2 == 0:
        pares_for.add(numero)

print("Números pares usando for:", pares_for)

# Buscar los números pares usando list comprehension
pares_comprehension = {numero for numero in numeros if numero % 2 == 0}
print("Números pares usando list comprehension:", pares_comprehension)

#Halle las permutaciones de un conjunto tomando dos elementos

letras = {'a', 'b', 'c'}

permutaciones = set([(a,b) for a in letras for b in letras if a != b]) # Genera permutaciones de dos elementos sin repetición

print(permutaciones)



```
