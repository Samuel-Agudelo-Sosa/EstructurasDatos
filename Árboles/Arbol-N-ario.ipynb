```python
class TreeNode:
    def __init__(self, value):
        self.value = value
        self.hijos = []
```

```python
class Tree:
    def __init__(self, root):
        self.root = TreeNode(root)
    
    def search(self, value):
        def search_node_recursive(current_node, value):
            if current_node.value == value:
                return True
            for hijo in current_node.hijos:
                if search_node_recursive(hijo, value):
                    return True
            return False
        if self.root is None:
            return False
        return search_node_recursive(self.root, value)
    
    def getNode(self, value):
        def get_node_recursive(current_node, value):
            if current_node.value == value:
                return current_node
            for hijo in current_node.hijos:
                result = get_node_recursive(hijo, value)
                if result is not None:
                    return result
            return None
        if self.root is None:
            return None
        return get_node_recursive(self.root, value)

    def insertar(self, parent, value):
        parent_node = self.getNode(parent)
        if parent_node is None:
            raise ValueError("Parent node not found")
        new_node = TreeNode(value)
        parent_node.hijos.append(new_node)
    
    def eliminar(self, value):
        def eliminar_node_recursive(current_node, value):
            for i, hijo in enumerate(current_node.hijos):
                if hijo.value == value:
                    del current_node.hijos[i]
                    return True
                if eliminar_node_recursive(hijo, value):
                    return True
            return False
        if self.root is None:
            return False
        if self.root.value == value:
            self.root = None
            return True
        return eliminar_node_recursive(self.root, value)
    
    def recorrido_dfs(self): # Depth-First Search
        elementos = []
        def recorrido_dfs_recursive(current_node): 
            elementos.append(current_node.value)
            for hijo in current_node.hijos:
                recorrido_dfs_recursive(hijo)
        if self.root is not None:
            recorrido_dfs_recursive(self.root)
        return elementos


    def __str__(self):
        """Define cómo se imprime el árbol de forma visual usando print(objeto)"""
        if self.root is None:
            return "Árbol vacío"
            
        lineas = []

        # Otra función anidada para encapsular la lógica visual
        def construir_string_recursivo(nodo_actual, nivel):
            prefijo = "└── " if nivel > 0 else ""
            espacios = "    " * (nivel - 1) if nivel > 1 else ""
            
            # En lugar de hacer print(), guardamos la línea en una lista
            lineas.append(f"{espacios}{prefijo}{nodo_actual.value}")
            
            for hijo in nodo_actual.hijos:
                construir_string_recursivo(hijo, nivel + 1)

        # Lanzamos la recursión
        construir_string_recursivo(self.root, nivel=0)
        
        # Juntamos todas las líneas con saltos de línea (\n)
        return "\n".join(lineas)
        
```

```python
# ==========================================
#          PRUEBA DE FUNCIONAMIENTO
# ==========================================
if __name__ == "__main__":
    print("--- 1. Creando el árbol con la raíz ---")
    mi_arbol = Tree("Electronicos")
    print(f"Raíz creada con éxito.\n")

    print("--- 2. Insertando elementos ---")
    # Insertamos hijos en la raíz
    mi_arbol.insertar("Electronicos", "Laptops")
    mi_arbol.insertar("Electronicos", "Celulares")

    # Insertamos hijos en las subcarpetas
    mi_arbol.insertar("Laptops", "MacBook")
    mi_arbol.insertar("Laptops", "ThinkPad")
    mi_arbol.insertar("Celulares", "iPhone")
    
    # Intento de inserción fallida para probar control de errores
   # mi_arbol.insertar("MarcaInexistente", "ElementoHuérfano")
    print("Inserciones terminadas.\n")

    print("--- 3. Probando el método Mágico __str__ (Impresión Visual) ---")
    # Al hacer print al objeto, Python llama automáticamente a __str__
    print(mi_arbol)
    print()

    print("--- 4. Probando el Recorrido DFS ---")
    resultado_dfs = mi_arbol.recorrido_dfs()
    print("Lista DFS generada:", resultado_dfs)
    print()

    print("--- 5. Probando búsquedas específicas ---")
    nodo_encontrado = mi_arbol.getNode("ThinkPad")
    if nodo_encontrado:
        print(f"✅ Búsqueda exitosa: Se encontró el objeto Nodo con valor '{nodo_encontrado.value}'")
    else:
        print("❌ Error: No se encontró el nodo.")
        
    nodo_falso = mi_arbol.getNode("PlayStation")
    print(f"🔍 Buscando 'PlayStation': {'Encontrado' if nodo_falso else 'No existe (Correcto)'}")
```
