# Implementación Árbol Binario de Búsqueda:

Se añade la clase Nodo:

```python
class Node:
 
    #Clase que representa un nodo del Árbol Binario de Búsqueda.   
    def __init__(self, value):
        self.value = value
        self.left = None  # Hijo Izquierdo 
        self.right = None # Hijo Derecho 
```

Se añade la clase BST:

```python
class BST:
    # arbol binario de busqueda
    def __init__(self):
        self.root = None # raiz vacia al inicio

    def insert(self, value):
        def insertRecursive(current_node, value):
            if value > current_node.value: 
                if current_node.right is None:
                    current_node.right = Node(value)
                else:
                    insertRecursive(current_node.right, value)
            elif value < current_node.value:
                if current_node.left is None:
                    current_node.left = Node(value)
                else:
                    insertRecursive(current_node.left, value)
            else:
                print("No se aceptan duplicados")

        # si el arbol esta vacio, el nuevo nodo es la raiz
        if self.root is None:
            self.root = Node(value)
        else: 
            insertRecursive(self.root, value)

    def search(self, value):
        def searchRecursive(current_node, value):
            if current_node is None:
                return False
            if current_node.value == value:
                return True
            # buscamos por la izquierda o derecha segun el valor
            elif value < current_node.value:
                return searchRecursive(current_node.left, value)
            else:
                return searchRecursive(current_node.right, value)

        return searchRecursive(self.root, value)
    
    def delete(self, value):
        def deleteRecursive(current_node, value):
            if current_node is None:
                return None
            
            # navegamos para encontrar el nodo
            if value < current_node.value:
                current_node.left = deleteRecursive(current_node.left, value)
            elif value > current_node.value:
                current_node.right = deleteRecursive(current_node.right, value)
            else:
                # nodo encontrado. caso un hijo o ninguno
                if current_node.left is None:
                    return current_node.right
                elif current_node.right is None:
                    return current_node.left

                # caso 3: dos hijos. buscamos el sucesor
                succesor = current_node.right
                while succesor.left is not None:
                    succesor = succesor.left
                
                # reemplazamos valor y borramos el duplicado
                current_node.value = succesor.value
                current_node.right = deleteRecursive(current_node.right, succesor.value)
            
            return current_node
    
        self.root = deleteRecursive(self.root, value)

    def preorder(self):
        def preorder_Recursive(current_node):
            if current_node is not None:
                acc.append(current_node.value) # raiz primero
                preorder_Recursive(current_node.left)
                preorder_Recursive(current_node.right)

        acc = [] 
        preorder_Recursive(self.root)
        return acc 
    
    def inorder(self):
        def inorder_Recursive(current_node):
            if current_node is not None:
                inorder_Recursive(current_node.left)
                acc.append(current_node.value) # raiz en medio
                inorder_Recursive(current_node.right)

        acc = [] 
        inorder_Recursive(self.root)
        return acc
    
    def postorder(self):
        def postorder_Recursive(current_node):
            if current_node is not None:
                postorder_Recursive(current_node.left)
                postorder_Recursive(current_node.right)
                acc.append(current_node.value) # raiz al final

        acc = [] 
        postorder_Recursive(self.root)
        return acc
```

# Análisis de rendimiento (Benchmarking):

Se implementan las pruebas:

```python
import time
import sys

sys.setrecursionlimit(2500) # por si acaso con el arbol degenerado

# 1. Prueba Arbol Degenerado
arbol_malo = BST()
for i in range(1000):
    arbol_malo.insert(i)

inicio = time.perf_counter()
arbol_malo.search(999) # buscamos el ultimo
fin = time.perf_counter()
print(f"Tiempo degenerado: {(fin - inicio) * 1000:.4f} ms")

# 2. Prueba Arbol Balanceado (No degenerado)
arbol_bueno = BST()
# insertamos los mismos 1000 numeros pero de forma que quede equilibrado
def insertar_mitades(inicio, fin):
    if inicio <= fin:
        mitad = (inicio + fin) // 2
        arbol_bueno.insert(mitad)
        insertar_mitades(inicio, mitad - 1)
        insertar_mitades(mitad + 1, fin)

insertar_mitades(0, 999)

inicio = time.perf_counter()
arbol_bueno.search(999)
fin = time.perf_counter()
print(f"Tiempo balanceado: {(fin - inicio) * 1000:.4f} ms")
```
