```python
def algoritmo(n):
    cnt = 0
    p = 1
    cnt += 1
    i = 1
    cnt += 1
    while i <= n/2:
        cnt += 1
        i = i + 1
        cnt += 1
        j = 1
        cnt += 1
        while j <= 2*n:
            cnt += 1
            j *= 2
            cnt += 1
            p += 1
            cnt += 1
        cnt += 1
    cnt += 1

    return (cnt,p)
```

```python
import math
def complejidadTeorica(n):
    l1 = 1
    l2 = 1
    l3 = math.floor(n/2) + 1
    l4 = math.floor(n/2)
    l5 = math.floor(n/2)
    l6 = (math.floor(math.log2(n)) + 3) * math.floor(n/2)
    l7 = (math.floor(math.log2(n)) + 2) * math.floor(n/2)
    l8 = l7
    p = l8 + 1
    return (l1 + l2 + l3 + l4 + l5 + l6 + l7 + l8, p) 
```

```python
for i in range(1,11):
    print(f"n:{i}, complejidadReal: {algoritmo(i)}, complejidadTeórica: {complejidadTeorica(i)} ")
```
